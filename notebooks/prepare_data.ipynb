```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../data/raw/train_ship_segmentations_v2.csv")
df.head()
```

```python
image_df = (
    df.groupby("ImageId")
      .agg(
          num_ships=("EncodedPixels", lambda x: x.notna().sum())
      )
      .reset_index()
)

image_df.head()
```

```python
image_df['is_ships'] = image_df['num_ships'] > 0

image_df.head()
```

```python
empty_images = image_df[image_df['num_ships'] == 0]
empty_images.head()
```

```python
ships_images = image_df[image_df['num_ships'] != 0]
ships_images.head()
```

```python
train_df, val_df = train_test_split(image_df, test_size=0.2, stratify=image_df['is_ships'], random_state=42, shuffle=True)
```
